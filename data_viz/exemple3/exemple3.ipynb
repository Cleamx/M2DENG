```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, np.pi, 50)
fig, ax = plt.subplots(2, 2)

ax[0, 0].plot(x, np.sin(x))
ax[0, 1].plot(x, np.cos(x))
ax[1, 0].plot(x, np.tan(x))
ax[1, 1].plot(x, np.exp(x))

plt.tight_layout()
plt.show()
```

```python
categories = ['category 1', 'category 2', 'category 3', 'category 4']
values = [80, 20, 50, 30]
fig, ax1 = plt.subplots()
ax1.bar(categories, values, color='skyblue')
ax1.set_ylabel('Values', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
ax2 = ax1.twinx()
cumulative_values = [sum(values[:i+1]) for i in range(len(values))]
ax2.plot(categories, cumulative_values, color='red', marker='o')
ax2.set_ylabel('Cumulative Values', color='red')
ax2.tick_params(axis='y', labelcolor='red')
plt.title('Pareto chart')
plt.show()
```

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    'Distribution 1': np.random.randn(100),
    'Distribution 2': np.random.randn(100) * 2,
    'Distribution 3': np.random.randn(100) * 3
})

sns.boxplot(data=df)
plt.title('Box plot for three distributions')
plt.show()
```

```python
data = np.random.randn(100), np.random.randn(100) * 2, np.random.randn(100) * 3


sns.violinplot(data=data, inner=None)
sns.swarmplot(data=data, color='black', alpha=0.5)


plt.title('Violin plot')
plt.show()
```

```python
languages = ['Python', 'JavaScript', 'TypeScript', 'Java', 'C#']
rankings = [[8, 6, 5, 3, 2, 1], [1, 2, 2, 2, 3, 2], [10, 9, 8, 5, 5, 3], [2, 3, 3, 4, 4, 4], [5, 4, 4, 6, 6, 5]]
colors = ['lime', 'magenta', 'purple', 'orange', 'cyan']

plt.figure(figsize=(10, 6))
for i, language in enumerate(languages):
    plt.plot(range(1, 7), rankings[i], marker='o', label=language, color=colors[i])
plt.gca().invert_yaxis()
plt.xticks(range(1, 7), [2018, 2019, 2020, 2021, 2022, 2023])
plt.yticks(range(1, 11))
plt.xlabel('Year')
plt.ylabel('Ranking')
plt.title('Programming Language Rankings Over Years')
plt.legend()
plt.grid(True)
plt.show()
```
